###5. Cleaning Categorical Variables
Categorical columns like Payment_of_Min_Amount contain values like 'Yes', 'No', and 'NM'. Type_of_Loan contains comma-separated lists with extra spaces or 'No Data'.

````python
# Cleaning Categorical Variables

# Strip leading/trailing whitespaces and standardize text case

cat_cols = df.select_dtypes(include=['object']).columns     # Selecting columns containing text or object-type data
for col in cat_cols:
  df[col] = df[col].astype(str).str.strip().str.title()     # .astype(str): Converts the values to strings (text)
                                                            # .str.strip(): Removes spaces from the beginning and end of each value
                                                            # .str.title(): Capitalizes the first letter of each word

# Clean specific categorical values
df['Payment_of_Min_Amount'] = df['Payment_of_Min_Amount'].replace({'Nm': 'Not Mentioned'})
````

###6. Checking for Outliers in Numerical Variables
Using the Interquartile Range (IQR) method helps detect extreme financial outliers in columns like Annual_Income and Outstanding_Debt.

````python
# Checking for Outliers in Numerical Variables

# IQR method to detect outliers in Annual_Income

Q1 = df['Annual_Income'].quantile(0.25)
Q3 = df['Annual_Income'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Annual_Income'] < lower_bound) | (df['Annual_Income'] > upper_bound)]
print(f"Number of outliers in Annual_Income: {len(outliers)}")
````

````python
# IQR method to detect outliers in Outstanding_Debt

Q1 = df['Outstanding_Debt'].quantile(0.25)
Q3 = df['Outstanding_Debt'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Outstanding_Debt'] < lower_bound) | (df['Outstanding_Debt'] > upper_bound)]
print(f"Number of outliers in Outstanding_Debt: {len(outliers)}")
````

####**Important Note**: We can handle Outliers by capping or clipping them. This Capping/Clipping of outliers is called **Winsorization** which lets us to retain data rows without letting extreme values skew models.

df['Outstanding_Debt'] = df['Outstanding_Debt'].clip(lower=lower_bound, upper=upper_bound)

````python
# IQR method to detect outliers in Credit_Utilization_Ratio

Q1 = df['Credit_Utilization_Ratio'].quantile(0.25)
Q3 = df['Credit_Utilization_Ratio'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Credit_Utilization_Ratio'] < lower_bound) | (df['Credit_Utilization_Ratio'] > upper_bound)]
print(f"Number of outliers in Credit_Utilization_Ratio: {len(outliers)}")
````

````python
# IQR method to detect outliers in Delay_from_due_date

Q1 = df['Delay_from_due_date'].quantile(0.25)
Q3 = df['Delay_from_due_date'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Delay_from_due_date'] < lower_bound) | (df['Delay_from_due_date'] > upper_bound)]
print(f"Number of outliers in Delay_from_due_date: {len(outliers)}")
````

###7. Using apply to Create a New Column
We can create a new feature, such as Annual_Savings, by subtracting estimated annual expenses or calculating net disposable income using .apply().

````python
# Create a custom function to categorize income level
def classify_income_tier(income):     # def is used to create a function
    if income < 25000:
        return 'Low Income'
    elif income < 75000:
        return 'Middle Income'
    else:
        return 'High Income'

# Apply function to create 'Income_Tier' column
df['Income_Tier'] = df['Annual_Income'].apply(classify_income_tier)     # .apply() applies your function to each value in that column
print(df['Income_Tier'].value_counts())
````

#### Chart - 1: Target Variable Distribution (credit_score) — Countplot / Bar Chart (Univariate Analysis)

````python
# Chart - 1 visualization code
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual aesthetic
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))          # figsize=(width, height)

# Order categories logically
order = ["Good", "Standard", "Poor"]

# Assign colors to categories
palette = {"Good": "green", "Standard": "yellow", "Poor": "red"}

# Create count plot (count plot counts how many times each category appears in your dataset)
ax = sns.countplot(
    data=df,
    x="credit_score",       # Displays credit score categories on the X-axis.
    order=order,
    hue="credit_score",     # Assigns colors based on the credit score categories.
    hue_order=order,        # Keeps the colors in the same logical order: Good, Standard, Poor.
    palette=palette,        # Assigns green, orange, and red to the categories.
    legend=False            # Hides the legend because the X-axis already displays the category names.
)

# Annotate counts and percentages on bars
total = len(df)

for p in ax.patches:            # ax.patches contains the rectangular bars in your count plot.
  height = p.get_height()
  percentage = f"{100 * height / total:.1f}%"     # This calculates the percentage of customers in the current category.

  ax.annotate(                                    # ax.annotate() is used to display the labels inside the bars
      f"{int(height):,}\n({percentage})",         # int(height) converts the bar height into an integer
                                                  # :, Adds commas to large numbers
                                                  # \n Moves the percentage to the next line
                                                  # ({percentage})	Displays the percentage in brackets.
      (p.get_x() + p.get_width() / 2.0, height / 2),  # p.get_x() gives the left position of the bar
                                                      # p.get_width() gives the width of the bar
                                                      # Dividing the width by 2 gives the midpoint
                                                      # height/2 places the text halfway up the bar
      ha="center",            # ha = horizontal alignment
      va="center",            # va = vertical alignment
      fontsize=11,
      color="white",
      fontweight="bold",
  )

# Add chart title and axis labels
plt.title("Distribution of Credit Score Categories", fontsize=14, weight="bold")
plt.xlabel("Credit Score Tier", fontsize=12)
plt.ylabel("Customer Count", fontsize=12)
plt.tight_layout()                # automatically adjusts the spacing around the chart
plt.show()
````
